```python
#1.Write a program to check if a number entered by the user is even or odd.
import random
def EVEN_ODD():
    nums=random.randint(1,100)
    if nums%2==0:
        print(f'the number is even {nums}')
    else:
        print(f'the number is odd {nums}')

EVEN_ODD()
```

```python
#2.Write a program to check whether a number is positive, negative, or zero.
num=eval(input("Enter the number:"))
if num>0:
    print(f'{num} number is positive')
elif num<0:
    print(f'{num} number is negitive')
elif num==0:
    print(f'{num} number is zero')
else:
    print(f'The {num} number is Invalid Number')
```
