```python
#1. Ramesh’s basic salary is input through the keyboard. His dearness allowance is 40% of basic salary, and 
# house rent allowance is 20% of basic salary. Write a program to calculate his gross salary.
#basicsalary=10000
#dear_allow=40
#Hra=20
da=eval(input("enter the da value:"))
hra=eval(input("enter the hra value:"))
basic_salary=eval(input("Enter the basic salary is:"))
da=0.40*basic_salary
hra=0.20*basic_salary
gross_salary=da+hra+basic_salary
gross_salary
```

```python
##### 2. The distance between two cities (in km.) is input through the keyboard.
#Write a program to convert and print this distance in meters, feet, inches and centimeters. 
#distane=510
#meters = km * 1000
#centimeters = km * 100000
#feet = km * 3280.84
#inches = km * 39370.1
km = float(input("Enter the distance in kilometers: "))
meters_per_km=1000
centimeters_per_km=100000
feet_per_km=3280.84
inches_per_km=39370.1
meters = km * meters_per_km
centimeters = km * centimeters_per_km
feet = km * feet_per_km
inches = km * inches_per_km
print(f"{km} km is equal to:")
print(f"{meters} meters")
print(f"{centimeters} centimeters")
print(f"{feet} feet")
print(f"{inches} inches")
```

```python
#3. If the marks obtained by a student in five different subjects are input through the keyboard, find out the
#aggregate marks and percentage marks obtained by the student. Assume that the maximum marks that
#can be obtained by a student in each subject is 100.
# Input marks for five subjects
Math = eval(input("Enter marks for Math: "))
English = eval(input("Enter marks for English: "))
Social_science = eval(input("Enter marks for Social_science: "))
Python=eval(input("Enter marks for Python: "))
Data_science=eval(input("Enter marks for Data_science: "))
aggregate =Math+English+Social_science+Python+Data_science
percentage = (aggregate / 500) * 100
print("Aggregate marks:", aggregate)
print("Percentage marks:", percentage, "%")
```

```python
#4. Temperature of a city in Fahrenheit degrees is input through the keyboard. Write a program to convert this
#temperature into Centigrade degrees.
# Input temperature in Fahrenheit
fahrenheit =eval(input("Enter temperature in Fahrenheit: "))
celsius = (fahrenheit - 32) * 5 / 9
# Display result
print("Temperature in Celsius:", celsius)
```

```python
#5. The length & breadth of a rectangle and radius of a circle are input through the keyboard. Write a program
#to calculate the area & perimeter of the rectangle, and the area & circumference of the circle.
length=eval(input("Enter the Length value:"))
Breadth=eval(input("Enter the Breadth value:"))
Radius=eval(input("Enter the Radius value:"))
Pi=eval(input("Enter the Pi value:"))
area_of_rectangle=length*Breadth
area_of_perimeter=(1/2)*(length+Breadth)
area_of_circle=Pi*Radius*Radius
circumference_of_circle=2*Pi*Radius
print("\nRectangle:")
print("area:",area_of_rectangle)
print("perimeter:",area_of_perimeter)
print("\nCircle:") 
print("area:",area_of_circle)
print("circumference:",circumference_of_circle)
```

```python
#6. Two numbers are input through the keyboard into two locations C and D. Write a program to interchange
#the contents of C and D.
a=eval(input("Enter the value for a:"))
b=eval(input("Enter the value fot b:"))
temp=a
a=b
b=temp
print("After swapping:")
print("a=:",a)
print("b=:",b)
```

```python
#7. If a five-digit number is input through the keyboard, write a program to reverse the number.
number=input("Enter the five-digit number:")
reverse=(number[::-1])
print("after reverse:",reverse)
```

```python
#8. If a four-digit number is input through the keyboard, write a program to obtain the sum of the first and last
#digit of this number.
#num=eval(input("Enter the four-digit number:"))
#first_digit=num//100
#last_digit=num%10
#sum_digit=first_digit+last_digit
#print("sum of first digit and last digit",sum_digit)
num=eval(input("Enter the four-digit number:"))
first_digit=num//1000 #Removes last 3 digits,Gives first digit
last_digit=num%10 #Gets remainder after dividing by 10,Gives last digit
sum_digit=first_digit+last_digit
print("sum of first digit and last digit",sum_digit)
```
