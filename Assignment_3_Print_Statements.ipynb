```python
# WAP calculate area of circle when radius given as 20
# assume pi=3.14  
# Formula = pi*r*r
pi=3.14
radius=20
area_of_circle=pi*radius*radius
print('the multiplication of 3.14 and 20 is 1256')
print('the multiplication of {} and {} is {}'.format(3.14,20,area_of_circle))
print('the multiplication of 3.14 and 20 is 1256')
print(f'the multiplication of {pi} and {radius} is {area_of_circle}')
```

```python
# WAP calculate area of right angle traingle when
# height given as 20, breadth given as 50
# Formual= 1/2*breadth*height
height=20
breadth=50
area_of_triangle=1/2*breadth*height
print('the multiplication of 0.5 and 50 and 20 is 500')
print('the multiplication of {} and {} and {} is {}')
print('the multiplication of {} and {} and {} is {}'.format(0.5,50,20,area_of_triangle))
print('the multiplication of 0.5 and 50 and 20 is 500')
print(f'the multiplication of {0.5} and {breadth} and {height} is {area_of_triangle}')
```
