```python
def sum_n(n):
    if n==1:
        return(1)
    return n+sum_n(n-1)

def factorial(n):
    if n==1:
        return(1)
    return n*factorial(n-1)

print('sum:',sum_n(5))
print('factorial:',factorial(5))
```
