**Try , except and finally**

```python
# try:working code should be here
# except : Error will catch here
#finally: wheather code works or fail
#         this blocks will run 
```

```python
try:
    print('coffee good')
except Exception as e:
    print('coffe not good')
finally:
    print('off the stove')
```
