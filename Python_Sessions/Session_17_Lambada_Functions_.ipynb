**Case:3 Function With if and else**

```python
def Max(a,b):
    if a>b:
        return(a)
    else:
        return(b)
Max(10,20)
```

```python
[<if out> <if con> else <else out> <for loop>]

lambda <arg> : <return>
lambda <arg1>,<arg2> : <return>
lambda <arg1>,<arg2> : <if out> <if con> else <else out>
```

```python
Max = lambda a,b : a if a>b else b
Max(10,20)
```

```python
Even_odd = lambda a : a if a%2 else a
Even_odd(5)
```

```python
G1 = lambda x : f'{X}_grade_A' if x>=75 else f'{x}_Grade_B' if x>=60 else f'{x}_Grade_c'
G1(10)
```

```python
G1 = lambda x : (f'{X}_grade_A' if x>=75 else\
                f'{x}_Grade_B' if x>=60 else\
                f'{x}_Grade_c')
G1(10)
```

```python
G2 = lambda x : (f'{X}_grade_A' if x>=75 else
                f'{x}_Grade_B' if x>=60 else
                f'{x}_Grade_c')
G2(68)
```
