```python
#calulate program
#we will take two number
#n1+2,n1-n2,n1/n2,n1*n2
#i want to pick one operation based on one condition
n1=eval(input("Enter the number 1:"))
n2=eval(input("Enter the number 2:"))
operation=eval(input("Enter the operation no:"))
if operation==1:
    print(n1+n2)
elif operation==2:
    print(n1-n2)
elif operation==3:
    print(n1/n2)
elif operation==4:
    print(n1*n2)
else:
    print("Enter valid operation")
```
