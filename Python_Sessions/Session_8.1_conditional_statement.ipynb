- so else statement does not need any condtion

```python
user_travel=eval(input("enter the user distance:"))
govt_travel=eval(input("enter the govt distance:"))
if user_travel>govt_travel:
    charge=eval(input('enter the charge perkm:'))
    update_distance=user_travel-govt_travel
    print(f'you need to pay charge for {update_distance}km')
    total_charge=update_distance*charge
    print(f'you need to pay charge {total_charge}')
else:
    print("enjoy a freeride")
```
