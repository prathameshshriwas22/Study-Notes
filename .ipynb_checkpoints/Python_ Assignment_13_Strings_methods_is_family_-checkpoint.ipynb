***istitle***

```python
str1='SHAH' #Return True if the string is a title-cased string, False otherwise.
#In a title-cased string, upper- and title-case characters may only
#follow uncased characters and lowercase characters only cased ones.
str1.istitle()
```

```python
bharat="Sanju Samson" # only the first letter of string should capital then print True
bharat.istitle()
```
