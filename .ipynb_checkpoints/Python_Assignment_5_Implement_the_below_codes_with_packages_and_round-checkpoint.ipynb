# find the area of circle

```python
# wap ask the user take the radius find the area of circle
# pi=3.14
# r=100
# area_of_circle=pi*r*r
# Example take pi value from math package
# r value using random
import random
import math
pi=math.pi
radius=random.randint(1,50)
area_of_circle=round(pi*radius*radius,2)
area_of_circle
```

# area of traingle

```python
# wap ask the user take height and breadth find the area of traingle
#       area_of_traingle=0.5*b*h
import random
height=random.randint(1,20)
breadth=random.randint(1,100)
area_of_triangle=0.5*height*breadth
area_of_triangle
```

# find the average

```python
# wap ask the user take 3 numbers find the average
import random
a=random.randint(1,10)
b=random.randint(2,20)
c=random.randint(3,30)
d=round(a+b+c/3,1)
d
```

# calculate total bill

# calculate tax amount

```python
# wap ask the user take the salary, percetage of tax calculate tax amount
import random
salary=random.randint(1000,5000)
tax=random.randint(1,20)
tax_amount=round(salary*tax/100,1)
tax_amount
```

# dollar convert into rupees
