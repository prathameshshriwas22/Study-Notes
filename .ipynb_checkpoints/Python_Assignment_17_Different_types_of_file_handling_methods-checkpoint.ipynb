```python
#Example 1) 'x' create a new file and open it for writing

file = open("student.txt", "x")
file.write("Rahul 85")
file.close()
```

```python
# Example 2) 'a' open for writing, appending to the end of the file if it exists

file = open("student.txt","a")
file.write("Vaibhav 87, Roshan 88, Gopal 89 , Avinash 93")
file.close()
```

```python
# Example 3) 'b' binary mode
file = open("student.txt", "wb")  # use w+b for binary mode write and binary
file.write(b"87,88,89,93")
file.close()
```

```python
# Example 4)'t' text mode (default)
#rt means:
#r → read
#t → text mode


file = open("student.txt", "rt")
content = file.read()
print(content)
file.close()
```

```python
# Example 5) '+' open a disk file for updating (reading and writing)

file = open("student.txt", "r+")
content = file.read()
print(content)

file.write("\nHello Good Morning")
file.close()
```
