```python
# Q1) [2,3,30,40,50,6,7]
#    find the maximum number in a list with out using max function
```

```python
l1=[2,3,30,40,50,6,7]
max_num=l1[0]
for i in l1:
    if i > max_num:
        max_num = i
print("maximum_number:",max_num)
```

```python
# Q2) [2,3,30,40,50,6,7]
#    find the second maximum number in a list with out using max function
```

```python
l1=[2,3,30,40,50,6,7]
largest = l1[0]
second_largest = l1[0]
for i in l1:
    if i > largest:
        second_largest = largest
        largest = i
    elif i > second_largest and i != largest:
        second_largest=i
print(second_largest)
```

```python
# Q3) [2,3,30,40,50,6,7]
#     sort the values with out sort method
```

```python
l1=[2,3,30,40,50,6,7]
for i in range(len(l1)):
    for j in range(i+1, len(l1)):
        if l1[i] > l1[j]:
            temp = l1[i]
            l1[i] = l1[j]
            l1[j] = temp
l1
```

```python
# Q4) [2,3,30,40,50,6,7]
#     reverse the values with out reverse method
```

```python
l1=[2,3,30,40,50,6,7]
start=0
end=len(l1)-1
while start < end:
    temp = l1[start]
    l1[start] = l1[end]
    l1[end] = temp

    start = start + 1
    end  = end - 1
l1
```

```python
# Q5) [2,3,[30,40,50],[6,7]] output will be = [2,3,30,40,50,6,7]
```

```python
l1=[2,3,[30,40,50],[6,7]]
n=len(l1)
output=[]
for i in l1:
    if type(i) == list:
        output.extend(i)
    else:
        output.append(i)

output
```

```python
# Q6) str1='ola ola ola ola'
#     find the number of 'ola' with out using count method
```

```python
str1='ola ola ola ola'

words = str1.split()
count = 0
for i in words:
    if i == 'ola':
        count=count+1
count
```

```python
# Q7) [10,20,30,10,10]
#     find the number of 10 with out using count
```

```python
l1=[10,20,30,10,10]
c=0
for i in l1:
    if i == 10:
        c+=1
c
```

```python
# Q8) str1='hello how are you, hello im good'
#     take any sentence like above find the most reptaed word
```

```python
str1='hello how are you, hello im good'

words = str1.split()

max_count = 0
most_word = ""

for i in words:
    c = 0
    for j in words:
        if i == j:
            c=c+1
    if c > max_count:
        max_count = c
        most_word = i

print("Most repeated Word:",most_word)
print("count:",max_count)
```
