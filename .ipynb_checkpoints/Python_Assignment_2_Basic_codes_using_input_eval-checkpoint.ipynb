```python
# WAP take three numbers find the average

# WAP calculate area of circle when radius given as 20
# assume pi=3.14  
# Formula = pi*r*r

# WAP calculate area of right angle traingle when
# height given as 20, breadth given as 50
# Formual= 1/2*breadth*height

# Wap Calculate area of rectangle
# Formualr= length*breadth

# Wap calculate the weight in kgs when weight in pounds
# one pound= 2.2kgs

# Wap calculate the money in rupees when money in dollars
# one dollar= 90rs

# Wap ask the user calculate tax amount for a given salary 100000 and tax percenatge is 10
```

```python
# WAP take three numbers find the average
a=eval(input('enter the number1:'))
b=eval(input('enter the number2:'))
c=eval(input('enter the number3:'))
avg=round((a+b+c)/3,1)
avg
```

```python
# WAP calculate area of circle when radius given as 20
# assume pi=3.14  
# Formula = pi*r*r
pi=eval(input('enter pi value:'))
radius=eval(input('enter radius:'))
area_of_circle=round(pi*radius*radius,1)
area_of_circle
```

```python
# WAP calculate area of right angle traingle when
# height given as 20, breadth given as 50
# Formual= 1/2*breadth*height
height=eval(input('enter the height:'))
breadth=eval(input('enter the breadth:'))
area_of_triangle=round(1/2*breadth*height,1)
area_of_triangle
```

```python
# Wap Calculate area of rectangle
# Formualr= length*breadth
length=eval(input('enter the length:'))
breadth=eval(input('enter the breadth:'))
area_of_rectangle=round(length*breadth,1)
area_of_rectangle
```

```python
# Wap calculate the weight in kgs when weight in pounds
# one pound= 2.2kgs
pounds_weight_in_kgs=eval(input('enter the pounds_weight_in_kgs:'))
one_pound=eval(input('enter the one pound:'))
weight_in_pounds=round(one_pound*pounds_weight_in_kgs,1)
weight_in_pounds
```

```python
# Wap calculate the money in rupees when money in dollars
# one dollar= 90rs
one_dollar=eval(input('one dollar:'))
money_in_dollar=eval(input('money in dollar:'))
money_in_rupees=round(one_dollar*money_in_dollar)
money_in_rupees
```

```python
# Wap ask the user calculate tax amount for a given salary 100000 and tax percenatge is 10
salary=eval(input('Enter the Value:'))
tax_per=eval(input('Enter the value:'))
tax_amount=round(salary*tax_per/100,1)
tax_amount
```
