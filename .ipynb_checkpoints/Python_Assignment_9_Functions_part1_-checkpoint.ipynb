```python
#Q1: Sum of Three Numbers
#Write a function `sum_three(a, b, c)` that returns the sum of the three numbers.
#Example:
#sum_three(3, 5, 7) ➞ 15
```

```python
def Sum_three(a,b,c):
    sum=a+b+c
    print(f'the sum of the {a},{b} and {c} is {sum}')
Sum_three(3,5,7)
```

```python
#Q2: Area of a Circle
#Write a function `area_circle(radius)` that returns the area of a circle. Use π = 3.14159
#Example:
#area_circle(7) ➞ 153.9381
```

```python

def area_of_circle(radius):
    pi=3.14
    area_of_circle=pi*radius*radius
    print(f'the area of circle is {area_of_circle}')

area_of_circle(7)
```

```python
#Q3: Check Even or Odd
#Write a function `check_even_odd(num)` that prints whether a number is Even or Odd.
#Example:
#check_even_odd(5) ➞ &quot;Odd&quot;
#check_even_odd(8) ➞ &quot;Even&quot;
```

```python
def check_even_odd(num):
    if num%2==0:
        print('The Number Is Even')
    else:
        print('The Number Is Odd')

check_even_odd(5)
check_even_odd(8)
```

```python
#Q4: Calculate BMI
#Write a function `calculate_bmi(weight, height)` that returns BMI value using the formula:
#BMI = weight / (height ** 2)
#Example:
#calculate_bmi(70, 1.75) ➞ 22.86
```

```python
def calculate_bmi(weight,height):
    BMI=weight/(height**2)
    print(f'the BMI value is {BMI}')
calculate_bmi(70,1.75)
```

```python
#Q6: Find Maximum of Two Numbers
#Write a function `find_max(a, b)` that returns the greater of the two numbers.
#Example:
#find_max(4, 9) ➞ 9
```

```python
def find_max(a,b):
    if a>b:
        print(f"{a} The number is greater number")
    elif a<b:
        print(f'{b} The number is greater number')
    else:
        print("Invalid Number")

find_max(10,20)
```

```python
#Q7: Power Function
#Write a function `power(base, exponent)` that returns base raised to the power exponent.
#Example:
#power(2, 3) ➞ 8
```

```python
def power(base,exponitinal):
    return base**exponitinal
power(2,3)
```

```python
#Q9: Check Leap Year
#Write a function `is_leap_year(year)` that returns True if the year is a leap year, otherwise
#False.
#Example:
#is_leap_year(2020) ➞ True
#is_leap_year(2023) ➞ False
```

```python
def Is_leap_year(year):
    if (year% 4 == 0 and year % 100!=0) or year % 400 == 0:
        print(f'True {year} is Leap Year')
    else:
        print(f'False {year} is Not Leap Year')

Is_leap_year(2020)
Is_leap_year(2023)
```
