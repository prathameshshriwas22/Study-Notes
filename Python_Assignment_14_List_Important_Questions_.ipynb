```python
# Q1) [2,3,30,40,50,6,7]
#    find the maximum number in a list with out using max function
```

```python
l1=[2,3,30,40,50,6,7]
max_num=l1[0]
for i in l1:
    if i > max_num:
        max_num = i
print("maximum_number:",max_num)
```

```python
# Q2) [2,3,30,40,50,6,7]
#    find the second maximum number in a list with out using max function
```

```python
l1=[2,3,30,40,50,6,7]
largest = l1[0]
second_largest = l1[0]
for i in l1:
    if i > largest:
        second_largest = largest
        largest = i
    elif i > second_largest and i != largest:
        second_largest=i
print(second_largest)
```

```python
# Q3) [2,3,30,40,50,6,7]
#     sort the values with out sort method
```

```python
l1=[2,3,30,40,50,6,7]
for i in range(len(l1)):
    for j in range(i+1, len(l1)):
        if l1[i] > l1[j]:
            temp = l1[i]
            l1[i] = l1[j]
            l1[j] = temp
l1
```

```python
# Q4) [2,3,30,40,50,6,7]
#     reverse the values with out reverse method
```

```python
l1=[2,3,30,40,50,6,7]
start=0
end=len(l1)-1
while start < end:
    temp = l1[start]
    l1[start] = l1[end]
    l1[end] = temp

    start = start + 1
    end  = end - 1
l1
```

```python
# Q8) str1='hello how are you, hello im good'
#     take any sentence like above find the most reptaed word
```

```python
str1='hello how are you, hello im good'

words = str1.split()

max_count = 0
most_word = ""

for i in words:
    c = 0
    for j in words:
        if i == j:
            c=c+1
    if c > max_count:
        max_count = c
        most_word = i

print("Most repeated Word:",most_word)
print("count:",max_count)
```

```python
# Q11) create counter program :
#      hello=2 how =1 are =2 you =1 im =1 good=1 ,=1
#      output should be alphabetical order
```

```python
str1 = 'hello how are you, hello are im good'

words = str1.split()

d = {}

for i in words:
    if i in d:
        d[i] = d[i] + 1
    else:
        d[i] = 1

for i in sorted(d):
    print(i, "=" ,d[i])
```

```python
# Q12) str1='The India is an Greatest Country!, Im ha$$$py here&&&'
#   out='India Greatest Country,Im hapy here'
```

```python
str1 = 'The India is an Greatest Country!, Im ha$$$py here&&&'

words = str1.split()

for i in words:
    w = ""
    for j in i:
        if j.isalpha():
            w = w + j
    if w not in ['The','is','an']:
        print(w,end=" ")
```

```python
# Q13) [202,301,405,607,590,320,140,50]   T
#      [102,201,305,407,390,220,140,50]   T
#      [102,201,305,407,390,220,440,50]  False
#    you need to create a function
#    the function should verify the above pattern with any list
```

```python
def check_pattern(l1):
    
    for i in l1:
        num = str(i)
        if num[0] != num[-1]:
            return False
        return True
    
print(check_pattern([202,301,405,607,590,320,140,50]))
print(check_pattern([102,201,305,407,390,220,140,50]))
print(check_pattern([102,201,305,407,390,220,440,50]))
```
