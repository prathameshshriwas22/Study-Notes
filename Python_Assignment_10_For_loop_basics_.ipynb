```python
# 1.Print all prime numbers between 1 and 100.

for num in range(2,101):
    if all(num%i !=0 for i in range(2,int(num**0.5)+1)):
        print(num,end=' ')
```

```python
# 2.Print all even and odds numbers between 1 and 200.

Evens=[num for num in range(1,201)if num%2 == 0]
Odds=[num for num in range(1,201) if num%2 != 0]
print("Even Numbers:")
print(Evens)
print("\nOdd Numbers:")
print(Odds)
```

```python
#Print the addition of numbers from a given number 
       #number (e.g., 374 → 3+7+4) using only loops.
number=374
total=0
#Loop until the number becomes 0
while number>0:
    digit = number % 10 # Get the last digit
    total += digit     # Add digit to total
    number //= 10       # Remove the last digit
print(total)
```

```python
# 4.Print the total length of the integer  number (e.g., 23789 → 5).
number=23789
length=0
while number>0:
    number //= 10  # Remove the last digit
    length += 1    # Increment the count

print(length)
```
