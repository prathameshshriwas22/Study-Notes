```python
# Q3) remove digits from string
# str1='hai123hello435'
# s='haihello'
```

```python
str1='hai123hello435'
s = ""
for char in str1:
    if char.isalpha():
        s = s + char
print(s)
```

```python
# Q5)replace vowels with '*'
# str1 = 'apple'
# s = '*ppl*'
```

```python
str1 = 'apple'
s = ""
for char in str1:
    if char in "aeiou":
        s = s + "*"
    else:
        s = s + char
print(s)
```

```python
# Q7)count the number of vowels in a string
# str1 = 'education'
# s = 5
```

```python
str1 = 'education'
count = 0
for char in str1:
    if char in "aeiou":
        count = count + 1
print(count)
```

```python
# Q8)check if a string is a palindrome
# str1 = 'madam'
# s = True
```

```python
str1 = 'madam'
if str1 == str1[::-1]:
    s = 'True'
else:
    s = 'False'
print(s)
```

```python
# Q9)remove all special characters from a string
# str1 = 'he@llo#wo$rld!'
# s = 'helloworld'
```

```python
str1 = 'he@llo#wo$rld!'
s = ""
for char in str1:
    if char.isalnum():
        s = s + char
print(s)
```

```python
# Q10)replace spaces with underscores
# str1 = 'machine learning is fun'
# s = 'machine_learning_is_fun'
```

```python
str1 = 'machine learning is fun'
s =""
for char in str1:
    if char == " ":
        s = s + "_"
    else:
        s = s + char
print(s)
```
